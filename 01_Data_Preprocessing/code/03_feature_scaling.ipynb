### Techniques Covered

- Standardization
- Min-Max Scaling
- Robust Scaling
- Normalization

```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import Normalizer
```

```python
df = pd.read_csv("../datasets/03_feature_scaling.csv")

df
```

# Standardization

Standardization transforms numerical features so that they have:

- Mean = 0
- Standard Deviation = 1

Formula:

Z = (X - Mean) / Standard Deviation

```python
standard_scaler = StandardScaler()

standard_scaled = standard_scaler.fit_transform(
    df[["age", "salary", "height", "weight"]]
)

standard_df = pd.DataFrame(
    standard_scaled,
    columns=["age", "salary", "height", "weight"]
)

standard_df
```

# Min-Max Scaling

Min-Max Scaling converts data into the range:

0 → 1

Formula:

(X - Min) / (Max - Min)

```python
minmax_scaler = MinMaxScaler()

minmax_scaled = minmax_scaler.fit_transform(
    df[["age", "salary", "height", "weight"]]
)

minmax_df = pd.DataFrame(
    minmax_scaled,
    columns=["age", "salary", "height", "weight"]
)

minmax_df
```

# Robust Scaling

Robust Scaling uses:

- Median
- Interquartile Range (IQR)

It is useful when the dataset contains outliers.

```python
robust_scaler = RobustScaler()

robust_scaled = robust_scaler.fit_transform(
    df[["age", "salary", "height", "weight"]]
)

robust_df = pd.DataFrame(
    robust_scaled,
    columns=["age", "salary", "height", "weight"]
)

robust_df
```

# Normalization

Normalization scales each row so that its length becomes 1.

It is commonly used in:

- Text Mining
- Cosine Similarity
- KNN

```python
normalizer = Normalizer()

normalized_data = normalizer.fit_transform(
    df[["age", "salary", "height", "weight"]]
)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=["age", "salary", "height", "weight"]
)

normalized_df
```

# Conclusion

In this notebook, we learned four Feature Scaling techniques:

- Standardization
- Min-Max Scaling
- Robust Scaling
- Normalization

Feature Scaling improves the performance of many Machine Learning algorithms by bringing numerical features to a comparable scale.
